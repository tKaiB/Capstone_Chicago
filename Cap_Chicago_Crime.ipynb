## Adding columns based on the current information acquired

```python
# Convert Date column to datetime format explicitly
df['Date'] = pd.to_datetime(df['Date'], format='%m/%d/%Y %I:%M:%S %p')

# Extract hour and bin into 4 categories:
# 1 (00:00 ~ 05:59), 2 (6:00 ~ 11:59), 3 (12:00 ~ 17:59), 4 (18:00 ~ 23:59)
df['Time_Of_Day'] = pd.cut(
    df['Date'].dt.hour,
    bins=[-1, 5, 11, 17, 24],
    labels=[1, 2, 3, 4]
).astype(int)
```

```python
# Extract the day name (e.g., Monday, Tuesday, etc.)
df['Day_of_Week'] = df['Date'].dt.day_name()
```

### --- Holiday Legend Mapping ---
- ID: 0 -> Non-Holiday
- ID: 1 -> New Year's Day
- ID: 2 -> Martin Luther King Jr. Day
- ID: 3 -> Lincoln's Birthday
- ID: 4 -> Washington's Birthday
- ID: 5 -> Casimir Pulaski Day
- ID: 6 -> Memorial Day
- ID: 7 -> Independence Day (observed)
- ID: 8 -> Independence Day

```python
import holidays

# Initialize Illinois/US holidays
us_holidays = holidays.US(state='IL')

# Extract unique dates in YYYY-MM-DD format to speed up holiday lookup
unique_dates = df['Date'].dt.date.unique()

# Create a mapping dictionary for holidays found in our dataset
holiday_mapping = {}
current_label = 1

# Assign 0 for non-holidays
holiday_mapping['Non-Holiday'] = 0

# Check each unique date and build our mapping
date_to_holiday_name = {}
for d in unique_dates:
    h_name = us_holidays.get(d)
    if h_name:
        # Clean holiday names slightly (e.g., handling Observed days if needed)
        if h_name not in holiday_mapping:
            holiday_mapping[h_name] = current_label
            current_label += 1
        date_to_holiday_name[d] = h_name
    else:
        date_to_holiday_name[d] = 'Non-Holiday'

# Create the 'Holiday' column by mapping the date to its holiday name and then to its number
df['Holiday_Name'] = df['Date'].dt.date.map(date_to_holiday_name)
df['Holiday'] = df['Holiday_Name'].map(holiday_mapping)

# Drop the temporary Holiday_Name helper column if you prefer to keep the dataset clean
df.drop(columns=['Holiday_Name'], inplace=True)

# Print the legend mapping numbers to holiday names
print("--- Holiday Legend Mapping ---")
for name, label in sorted(holiday_mapping.items(), key=lambda x: x[1]):
    print(f"ID: {label} -> {name}")

# Verify by showing some detected holidays
display(df[df['Holiday'] > 0][['Date', 'Holiday']].head())
```

#### Location Types

```python
import pandas as pd

# For Location Types
location_types_url = 'https://raw.githubusercontent.com/tKaiB/Capstone_Chicago/refs/heads/main/Location%20Types.csv'
## Location Types
location_types_df = pd.read_csv(location_types_url, skiprows=3, header=None)

# Transpose so each row becomes a record
location_types_df = location_types_df.T

# # Drop the first row (was the original row index 0, i.e., the "Locations:" header cell)
location_types_df = location_types_df.iloc[1:].reset_index(drop=True)

# # Rename columns for easy combining after
location_types_df.columns = ['Location_Type_ID', 'Location Description']


# Convert ID column to int to match your main df
location_types_df['Location_Type_ID'] = pd.to_numeric(location_types_df['Location_Type_ID'], errors='coerce').dropna().astype(int)

# Let's inspect their columns to make sure we map them correctly
print("Location Types Columns:", location_types_df.columns.tolist())

location_types_df
```
